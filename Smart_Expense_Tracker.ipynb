```python
# Personal Expense Tracker

print("================================")
print("     PERSONAL EXPENSE TRACKER")
print("================================")
print("Track your daily expenses easily!")
```

```python
expenses = []

print("Expense tracker is ready!")
```

```python
name = input("Enter expense name: ")
amount = float(input("Enter amount in rupees: "))

expenses.append([name, amount])

print("Expense added successfully!")
```

```python
print("----- All Expenses -----")

for expense in expenses:
    print(expense[0], ":", "₹", expense[1])
```

```python
total = 0

for expense in expenses:
    total = total + expense[1]

print("Total Expenses: ₹", total)
```

```python
with open("expenses.txt", "w", encoding="utf-8") as file:
    for expense in expenses:
        file.write(f"{expense[0]} - ₹{expense[1]}\n")

print("Expenses saved successfully!")
```

```python
with open("expenses.txt", "r", encoding="utf-8") as file:
    print(file.read())
```
